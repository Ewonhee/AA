```python
# 1. 파일을 열고, input 함수 자체를 파일의 한 줄을 읽어오는 기능으로 바꿔치기 합니다.
f = open("input.txt", "rt", encoding="utf-8")
input = f.readline
n, k = map(int, input().split())
cnt=0

for i in range(1, n+1):
    if n%i==0:
        cnt+=1
        if cnt==k:
            print(i)
            break
else:
    print(-1)
```
